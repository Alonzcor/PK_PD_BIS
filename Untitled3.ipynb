```python
import vitaldb
import pandas as pd
import numpy as np

# 1. Definir el paciente (caso 1) y los canales (tracks)
case_id = 1
tracks = ['BIS/EEG1_WAV', 'BIS/BIS']

# 2. Descargar los datos a 128 Hz (la frecuencia máxima del EEG)
# vitaldb alinea automáticamente los tracks a 128 Hz (intervalo = 1/128 s)
fs = 128
vals = vitaldb.load_case(case_id, tracks, interval=1/fs)

# 3. Construir vector de tiempo en segundos y DataFrame
tiempo = np.arange(len(vals)) / fs

df = pd.DataFrame({
    'Tiempo_s': tiempo,
    'EEG_uV': vals[:, 0],   # Señal cruda de ondas cerebrales (128 Hz)
    'BIS_real': vals[:, 1]  # Valor del monitor BIS (se repite cada 128 muestras o tiene NaNs)
})

# Rellenar los valores intermedios de BIS_real (forward-fill) para tener el valor vigente en cada instante
df['BIS_real'] = df['BIS_real'].ffill()

# 4. Guardar a CSV en el entorno de Colab
nombre_archivo = f'paciente_{case_id}_BIS.csv'
df.to_csv(nombre_archivo, index=False)

print(f"Archivo guardado: {nombre_archivo}")
print(f"Duración total de la cirugía: {tiempo[-1]/60:.1f} minutos")
print(df.dropna().head(10))
```

```python
from scipy.signal import butter, filtfilt
import matplotlib.pyplot as plt

# 1. Eliminar filas sin EEG o donde el monitor BIS aún no daba lectura válida (BIS <= 0)
df_clean = df.dropna(subset=['EEG_uV', 'BIS_real']).copy()
df_clean = df_clean[df_clean['BIS_real'] > 0].reset_index(drop=True)

# 2. Diseño de filtro Pasa-Banda Butterworth (0.5 Hz a 45 Hz) a fs = 128 Hz
fs = 128.0
lowcut = 0.5
highcut = 45.0
nyq = 0.5 * fs
b, a = butter(N=4, Wn=[lowcut / nyq, highcut / nyq], btype='band')

# Aplicar filtrado sin desfase (filtfilt)
df_clean['EEG_filtrado'] = filtfilt(b, a, df_clean['EEG_uV'])

print(f"Muestras válidas tras limpieza: {len(df_clean)} ({len(df_clean)/(fs*60):.1f} min)")
print(f"Rango de BIS real en esta cirugía: {df_clean['BIS_real'].min():.1f} a {df_clean['BIS_real'].max():.1f}")

# 3. Graficar la evolución del BIS real durante toda la cirugía
# Submuestreamos cada 128 puntos (1 Hz) solo para graficar rápido
plt.figure(figsize=(12, 4))
plt.plot(df_clean['Tiempo_s'][::128] / 60.0, df_clean['BIS_real'][::128], color='navy', linewidth=1.2)
plt.axhspan(40, 60, color='green', alpha=0.15, label='Zona objetivo anestesia general (40 - 60)')
plt.xlabel('Tiempo de cirugía (minutos)')
plt.ylabel('Índice BIS Real')
plt.title('Evolución de la Profundidad Anestésica - Paciente 1')
plt.ylim(0, 105)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
from scipy.signal import welch
from scipy.stats import entropy

# Configuración de ventanas de 4 segundos (512 muestras a 128 Hz) sin solapamiento
fs = 128
win_sec = 4
win_samples = fs * win_sec
n_windows = len(df_clean) // win_samples

features = []

for i in range(n_windows):
    start = i * win_samples
    end = start + win_samples

    seg_eeg = df_clean['EEG_filtrado'].iloc[start:end].values
    seg_bis = df_clean['BIS_real'].iloc[start:end].mean()
    seg_t = df_clean['Tiempo_s'].iloc[start:end].mean() / 60.0

    # Descartar ventanas con artefactos extremos (ej. electrocauterio > 100 uV)
    if np.max(np.abs(seg_eeg)) > 100:
        continue

    # Densidad Espectral de Potencia (PSD) con método de Welch
    freqs, psd = welch(seg_eeg, fs=fs, nperseg=fs*2)

    # Acotar espectro útil entre 0.5 y 45 Hz
    mask_total = (freqs >= 0.5) & (freqs <= 45.0)
    freqs_u = freqs[mask_total]
    psd_u = psd[mask_total]
    total_power = np.sum(psd_u)

    if total_power == 0:
        continue

    # 1. Potencia relativa Delta (0.5 - 4 Hz)
    p_delta = np.sum(psd_u[(freqs_u >= 0.5) & (freqs_u < 4.0)]) / total_power

    # 2. Potencia relativa Beta (13 - 30 Hz)
    p_beta = np.sum(psd_u[(freqs_u >= 13.0) & (freqs_u <= 30.0)]) / total_power

    # 3. SEF95 (Frecuencia al 95% de potencia acumulada)
    cum_power = np.cumsum(psd_u) / total_power
    sef95 = freqs_u[np.searchsorted(cum_power, 0.95)]

    # 4. Entropía Espectral normalizada
    psd_norm = psd_u / total_power
    spec_ent = entropy(psd_norm) / np.log(len(psd_norm))

    features.append([seg_t, p_delta, p_beta, sef95, spec_ent, seg_bis])

df_feat = pd.DataFrame(features, columns=['Tiempo_min', 'Rel_Delta', 'Rel_Beta', 'SEF95', 'Entropia', 'BIS_real'])

print(f"Ventanas limpias procesadas: {len(df_feat)}")
print("\nCorrelación de cada métrica con el BIS real:")
print(df_feat[['Rel_Delta', 'Rel_Beta', 'SEF95', 'Entropia', 'BIS_real']].corr()['BIS_real'])
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score

# 1. Suavizado temporal (ventana móvil de 4 bloques = 16 segundos, igual que el monitor real)
df_smooth = df_feat.copy()
cols_metrics = ['Rel_Delta', 'Rel_Beta', 'SEF95', 'Entropia']
for col in cols_metrics:
    df_smooth[col] = df_smooth[col].rolling(window=4, min_periods=1).mean()

# 2. Diagnóstico en consola: Comparar estado DESPIERTO (BIS > 80) vs ANESTESIA PROFUNDA (BIS < 40)
despierto = df_smooth[df_smooth['BIS_real'] > 80]
profundo = df_smooth[df_smooth['BIS_real'] < 40]

print("--- DIAGNÓSTICO NEUROFISIOLÓGICO (Valores Promedio) ---")
print(f"{'Métrica':<12} | {'Despierto (BIS>80)':<20} | {'Anestesia Profunda (BIS<40)':<25}")
print("-" * 65)
for col in cols_metrics:
    print(f"{col:<12} | {despierto[col].mean():<20.3f} | {profundo[col].mean():<25.3f}")

# 3. Entrenar modelo no lineal para predecir el BIS
X = df_smooth[cols_metrics]
y = df_smooth['BIS_real']

# Entrenamos usando los primeros 70% de la cirugía y probamos en el 30% final (despertar incluido)
split_idx = int(len(df_smooth) * 0.70)
X_train, X_test = X.iloc[:split_idx], X.iloc[split_idx:]
y_train, y_test = y.iloc[:split_idx], y.iloc[split_idx:]

modelo_bis = RandomForestRegressor(n_estimators=100, max_depth=8, random_state=42)
modelo_bis.fit(X_train, y_train)

# Predicción sobre toda la cirugía y evaluación en el tramo de prueba (no visto)
df_smooth['BIS_predicho'] = modelo_bis.predict(X)
mae_test = mean_absolute_error(y_test, df_smooth['BIS_predicho'].iloc[split_idx:])

print("\n--- RENDIMIENTO DEL ESTIMADOR DE BIS ---")
print(f"Error Medio Absoluto (MAE) en tramo de prueba: {mae_test:.2f} puntos BIS")
print("Importancia de cada métrica en el modelo:")
for name, imp in zip(cols_metrics, modelo_bis.feature_importances_):
    print(f"  - {name}: {imp*100:.1f}%")

# 4. Visualización en 2 paneles
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 7), sharex=True)

# Panel superior: Evolución del SEF95 (frecuencia borde) y Potencia Delta
ax1.plot(df_smooth['Tiempo_min'], df_smooth['SEF95'], color='teal', label='SEF95 (Hz) - Velocidad cerebral')
ax1.set_ylabel('Frecuencia SEF95 (Hz)', color='teal')
ax1.grid(True, alpha=0.3)
ax1_twin = ax1.twinx()
ax1_twin.plot(df_smooth['Tiempo_min'], df_smooth['Rel_Delta'], color='orange', alpha=0.7, label='Potencia Delta (%)')
ax1_twin.set_ylabel('Potencia Relativa Delta', color='orange')
ax1.set_title('Comportamiento de las Características Espectrales del EEG')

# Panel inferior: BIS Real vs BIS Estimado por nuestro algoritmo
ax2.plot(df_smooth['Tiempo_min'], df_smooth['BIS_real'], color='navy', linewidth=1.5, label='BIS Real (Monitor comercial)')
ax2.plot(df_smooth['Tiempo_min'], df_smooth['BIS_predicho'], color='crimson', linestyle='--', linewidth=1.3, label='BIS Estimado (Nuestro algoritmo)')
ax2.axvline(df_smooth['Tiempo_min'].iloc[split_idx], color='black', linestyle=':', label='Inicio tramo de prueba (30% final)')
ax2.axhspan(40, 60, color='green', alpha=0.12)
ax2.set_xlabel('Tiempo de cirugía (minutos)')
ax2.set_ylabel('Índice BIS')
ax2.set_title(f'Validación: BIS Real vs Estimado (MAE Prueba = {mae_test:.2f})')
ax2.set_ylim(0, 105)
ax2.grid(True, alpha=0.3)
ax2.legend(loc='lower right')

plt.tight_layout()
plt.show()
```
